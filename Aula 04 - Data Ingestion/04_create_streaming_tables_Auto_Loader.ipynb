## Cria a STREAMING TABLE usando Databricks Auto Loader

```sql
CREATE OR REFRESH STREAMING TABLE smart_claims_dev.`01_bronze`.claims
SCHEDULE EVERY 1 WEEK
AS
  SELECT
  *
  FROM read_files(
    '/Volumes/smart_claims_dev/00_landing/sql_server/claims.csv',
    format => 'csv',
    header => true
      
  )
```
